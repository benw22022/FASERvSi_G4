# Line intersections in 2D
https://stackoverflow.com/questions/563198/how-do-you-detect-where-two-line-segments-intersect

```python
def cross2d(v, w):
    return v[0] * w[1] - v[1] * w[0]
```

```python
q = np.array([-129.27, -14.6113])
s = np.array([-1.21965, -14.6113]) - q

p = np.array([-128.38, -10.7966]) 
r = np.array([-0.432093, -15.9172]) - p
```

```python
t = cross2d((q - p), s) / cross2d(r, s)
u = cross2d((q - p), r) / cross2d(r, s)
```

```python
print(f" t = {t}")
print(f" u = {u}")
```

```python
i = p + t * r
j = q + u * s
```

```python
print(f"i = {i}")
print(f"j = {j}")
```

```python
fig, ax = plt.subplots()
ax.plot([-129.27, -1.21965], [-14.6113, -14.6113]);
ax.plot([-128.38, -0.432093], [-10.7966, -15.9172]);
ax.scatter(i[0], i[1]);
```
